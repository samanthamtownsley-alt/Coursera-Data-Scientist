### Step 1: Import Required Libraries


```python
import pandas as pd
```

### Step 2: Load the Dataset Into a Dataframe


#### **Read Data**
<p>
The functions below will download the dataset into your browser:
</p>


```python
file_path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(file_path)

# Display the first few rows to ensure it loaded correctly
print(df.head())
```

### Step 3. Finding and Removing Duplicates
##### Task 1: Identify duplicate rows in the dataset.


```python
## Write your code here
# count number of duplicate rows in dataframe
duplicate_count = df.duplicated().sum()
print(f'The number of duplicate rows in the dataset is {duplicate_count}')
```

##### Task 2: Remove the duplicate rows from the dataframe.



```python
## Write your code here
#drop duplicates using drop_duplicates() function
df.drop_duplicates(inplace=True) #inplace=true keeps it in the same dataframe
duplicate_count = df.duplicated().sum()
print(f'The number of duplicate rows in the dataset is {duplicate_count}')
```

### Step 4: Finding Missing Values
##### Task 3: Find the missing values for all columns.


```python
## Write your code here
missing_values = df.isna().sum() #find number of missing values for all columns
print(missing_values.to_string())
```

##### Task 4: Find out how many rows are missing in the column RemoteWork.


```python
## Write your code here
missing_values = df['RemoteWork'].isna().sum() #find number of missing rows in column RemoteWork
print(f'The number of missing rows in column RemoteWork is {missing_values}')
```

### Step 5. Imputing Missing Values
##### Task 5: Find the value counts for the column RemoteWork.


```python
## Write your code here
df['RemoteWork'].value_counts() #find value counts for RemoteWork
```

##### Task 6: Identify the most frequent (majority) value in the RemoteWork column.



```python
## Write your code here
remotework_mode = df['RemoteWork'].mode()[0]
print(f'The most frequent value in RemoteWork is {remotework_mode}')
```

##### Task 7: Impute (replace) all the empty rows in the column RemoteWork with the majority value.



```python
## Write your code here
# fill all empty rows in column RemoteWork with majority values
df['RemoteWork'] = df['RemoteWork'].fillna(remotework_mode)
print(f'The number of missing rows in column RemoteWork is {df['RemoteWork'].isna().sum()}')
```

##### Task 8: Check for any compensation-related columns and describe their distribution.



```python
## Write your code here
# Analyze the three currency columns
cols_to_analyze = ['CompTotal', 'Currency', 'ConvertedCompYearly']
print(df[cols_to_analyze].describe(include='all'))
```
