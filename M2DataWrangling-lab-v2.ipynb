### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# count number of duplicate rows in dataframe
duplicate_count = df.duplicated().sum()
print(f'The number of duplicate rows in the dataset is {duplicate_count}')
#drop duplicates using drop_duplicates() function
df.drop_duplicates(inplace=True) #inplace=true keeps it in the same dataframe
duplicate_count = df.duplicated().sum()
print(f'The number of duplicate rows in the dataset is {duplicate_count}')

missing_values = df['Country'].isna().sum() #find number of missing rows in column Country
print(f'The original number of missing rows in column Country is {missing_values}')

clean_countries = df['Country'].dropna().unique()
countries_sorted = sorted(clean_countries) 
print(countries_sorted)

df['Country'] = df['Country'].fillna('Unknown')

#find updated number of missing rows in column Country
print(f'The number of missing rows in column Country is {df['Country'].isna().sum()} in updated df')
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
missing_values = df['EdLevel'].isna().sum() #find number of missing rows in column EdLevel
print(f'The original number of missing rows in column EdLevel is {missing_values}')

print(f'The different values for EdLeve are {df['EdLevel'].unique()}')

#Defining a dictionary for giving EdValues a consistent format
ed_mapping = {
    "Primary/elementary school": "Primary School",
    "Bachelor’s degree (B.A., B.S., B.Eng., etc.)": "Bachelor's Degree",
    "Master’s degree (M.A., M.S., M.Eng., MBA, etc.)": "Master's Degree",
    "Some college/university study without earning a degree": "Some College",
    "Secondary school (e.g. American high school, German Realschule or Gymnasium, etc.)": "High School",
    "Professional degree (JD, MD, Ph.D, Ed.D, etc.)": "Progessional Degree",
    "Associate degree (A.A., A.S., etc.)": "Associate Degree",
    "Something else": "Other",
}

# Apply the mapping to the column
# We use .replace() to swap the old strings for the new ones
df['EdLevel'] = df['EdLevel'].replace(ed_mapping)

# Replace nan with "unknown"
df['EdLevel'] = df['EdLevel'].fillna('Unknown')

# Verify the changes
print(df['EdLevel'].unique())

```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here

# Apply one-hot encoding to the Employment column
# We use prefix='Emp' to make the new column names clear (e.g., Emp_Full-time)
df_encoded = pd.get_dummies(df, columns=['Employment'], prefix='Emp')

# Display the first few rows of the new columns
# This filters the list of columns to only show the ones we just created
emp_cols = [col for col in df_encoded.columns if col.startswith('Emp_')]
print(df_encoded[emp_cols].head())

# Check the new dimensions (number of columns will increase)
print(f"Original columns: {df.shape[1]}")
print(f"New columns after encoding: {df_encoded.shape[1]}")
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here

#get the number of missing values for each column
missing_values = df.isnull().sum()

#sort the counts in descending order to see highest first
#Show top 10 columns with most missing values
highest_missing = missing_values.sort_values(ascending = False).head(10).index

#Show top 10 columns numbers of missing values and data type
summary_df = pd.DataFrame({
    'Missing Values': df[highest_missing].isnull().sum(),
    'Data Type': df[highest_missing ].dtypes
})
print(summary_df)
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here

#calculate the median
median_comp = df['ConvertedCompYearly'].median()

# Fill in missing values
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(median_comp)

print(f'Median ConvertedCompYearly: {median_comp}')

print(f'Missing values in ConvertedCopYearly; {df['ConvertedCompYearly'].isnull().sum()}')
```

```python
# Automatically create a list of columns where the data type is float64
num_cols = [col for col in df.columns if df[col].dtype == 'float64']

print(f"Numerical columns found: {num_cols}")

# Loop through the list and impute with the median
for col in num_cols:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)
    print(f"Imputed {col} with median: {median_val}")

# Final check for any remaining NaNs in those columns
print(f"Missing values after loop: {df[num_cols].isnull().sum()}")
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# Automatically create a list of columns where the data type is str
string_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()

print(f"String columns found: {len(string_cols)}")

# Loop through the list and impute with the mode
for col in string_cols:
# Check if the column is NOT entirely empty
    if not df[col].dropna().empty:
        mode_val = df[col].mode()[0]
        
        # Use .loc to force the update on the original dataframe
        df.loc[:, col] = df[col].fillna(mode_val)
        
        # print(f"Imputed {col} with mode: {mode_val}") # Optional: can be very long
    else:
        print(f"Skipping {col}: Column is entirely empty.")

# Final check for any remaining NaNs in those columns
print(f"Missing values after loop: {df[string_cols].isnull().sum()}")


```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here

#Calculate minimum and maximum values of ConvertedCompYearly
min_comp = df['ConvertedCompYearly'].min()
max_comp = df['ConvertedCompYearly'].max()
print(f'the original minimum comp is {min_comp} and the original maximum comp is {max_comp}')

df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - min_comp)/(max_comp - min_comp)
print(f'New normalized MinMax : {df['ConvertedCompYearly_MinMax'].min()} to {df['ConvertedCompYearly_MinMax'].max()}')
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

#log transform ConvertedCompYearly columng to reduce skewness
df['ConvertedCompYearly_Log'] = np.log1p(df['ConvertedCompYearly'])
print('new mAX to MIN for converted comp yearly log transform: ', df['ConvertedCompYearly_Log'].max(),',' , df['ConvertedCompYearly_Log'].min())
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here

#create new column `ExperienceLevel` based on the `YearsCodePro` column:
# Convert the 'YearsCodePro' column to numbers
df['YearsCodePro'] = df['YearsCodePro'].replace({'Less than 1 year': 0, 'More than 50 years': 51})
df['YearsCodePro'] = pd.to_numeric(df['YearsCodePro'], errors='coerce')

# 2. Define a function for the Experience Levels
def get_experience_level(years):
    if years < 2:
        return 'Junior'
    elif 2 <= years <= 5:
        return 'Intermediate'
    elif years > 5:
        return 'Senior'
    else:
        return 'Unknown'

# 3. Apply the function to create the new column
df['ExperienceLevel'] = df['YearsCodePro'].apply(get_experience_level)

# 4. Verify the distribution
print(df['ExperienceLevel'].value_counts())
```
